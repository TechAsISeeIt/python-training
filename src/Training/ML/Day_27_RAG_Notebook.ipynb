```python
docs = [
    "Python is a programming language",
    "Machine learning is used for predictions",
    "FastAPI is used to build APIs",
    "Transformers are used in NLP tasks",
]
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
doc_embeddings = model.encode(docs)
print(doc_embeddings)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

query = "What is Python?"
query_emb = model.encode([query])
scores = cosine_similarity(query_emb, doc_embeddings)[
    0
]  # it assigned indexes based on Similarity of cosine angles
top_idx = scores.argmax()  # it will give the index of the highest score
print(docs[top_idx])
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")

context = docs[top_idx]
"""
        Context: FastAPI is used to build APIs
        Question: What is used for building APIs?
        Answer:
"""
prompt = f"Context: {context}\nQuestion: {query}\nAnswer:"
result = generator(prompt, max_length=50)
print(result[0]["generated_text"])
```
