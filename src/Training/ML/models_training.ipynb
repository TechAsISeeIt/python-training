```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
import joblib
```

```python
df = pd.read_csv("student_ml_dataset.csv")

X = df[["hours_studied", "attendance"]]
y = df["pass"]

model = LogisticRegression()
model.fit(X, y)

joblib.dump(model, "model.pkl")
```
