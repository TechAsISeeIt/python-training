# Day 24 ML Training

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("student_ml_dataset.csv")
display(df)
```

```python
X = df[["hours_studied", "attendance"]]  # Select Feature Columns i.e Inputs or X
y = df["pass"]  # Select Label Columns i.e Outputs or Y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = LogisticRegression()  # Chose the model - Linear Models - LogisticRegression
model.fit(
    X_train, y_train
)  # Fit is method which will fit your model according to your training data
```

```python
pred = model.predict(
    X_test
)  # Predict Method used your model to test the data - Test inputs - X_test
accuracy_score(
    y_test, pred
)  # accuracy_score method used to evaluate matching percentage prediction to test outputs
```
